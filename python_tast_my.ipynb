```python
def skills(**skill):
    for key, value in skill.items():
        print(f"{key} → {value}")

skills(lang='Python', tool='SQL', soft='Teamwork')
```

ما هي مكتبة NumPy؟

NumPy اختصار لـ: Numerical Python.

مكتبة في بايثون متخصصة في:

التعامل مع المصفوفات Arrays (أسرع من القوائم list).

إجراء عمليات رياضية وإحصائية متقدمة.

الأساس لمعظم مكتبات الذكاء الاصطناعي وتحليل البيانات (مثل Pandas, Scikit-learn, TensorFlow).

```python
import numpy as np

# مصفوفة 1D
arr1 = np.array([1, 2, 3, 4])
print(arr1)

# مصفوفة 2D
arr2 = np.array([[1, 2, 3], [4, 5, 6]])
print(arr2)
```

🟢 دوال إنشاء سريعة

```python
print(np.zeros((2,3)))     # مصفوفة كلها أصفار
print(np.ones((3,3)))      # مصفوفة كلها واحدات
print(np.eye(3))           # مصفوفة هوية Identity Matrix
print(np.arange(0,10,2))   # مثل range() لكن لمصفوفة
print(np.linspace(0,1,5))  # أرقام متدرجة بين 0 و 1
```

🟢 العمليات الحسابية

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(a + b)   # جمع
print(a - b)   # طرح
print(a * b)   # ضرب عنصر بعنصر
print(a / b)   # قسمة
```

🟢 خصائص المصفوفة

```python
arr = np.array([[1,2,3],[4,5,6]])

print(arr.shape)   # أبعاد المصفوفة (صفوف × أعمدة)
print(arr.size)    # عدد العناصر
print(arr.ndim)    # عدد الأبعاد
print(arr.dtype)   # نوع البيانات
```

```python
arr = np.array([1, 2, 3, 4])

print(np.sum(arr))     # مجموع العناصر
print(np.mean(arr))    # المتوسط
print(np.std(arr))     # الانحراف المعياري
print(np.max(arr))     # أكبر قيمة
print(np.min(arr))     # أصغر قيمة
```

```python
# درجات طلاب في مادة
grades = np.array([60, 70, 80, 90, 100])

print("المجموع:", np.sum(grades))
print("المتوسط:", np.mean(grades))
print("أعلى درجة:", np.max(grades))
print("أقل درجة:", np.min(grades))
```
